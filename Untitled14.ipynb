```python
high_ANNUAL_RAINFALL = jk[jk[' ANNUAL RAINFALL'] > 3000]
```

```python
KK=high_ANNUAL_RAINFALL.groupby("YEAR")[" ANNUAL RAINFALL"].sum()
```

```python
KK
```

```python
plt.figure(figsize=(10,7))
plt.plot(KK.index,KK.values, marker='o', color='b', linestyle='-', linewidth=2)
plt.title('ANNUAL RAINFALL BY YEAR')
plt.xlabel('Year')
plt.ylabel('ANNUAL RAINFALL')
plt.grid(True)
plt.show()
```

### HERE THE GRAPH IS SHOWING ANNUAL RAIN FALL,AND WE CAN CLEARLY SEE THE VARIATIONS

```python
plt.figure(figsize = (12,6))
plt.title('Annual rainfall by year')
plt.xlabel('Year')
plt.ylabel('Annual Rainfall')
plt.bar(jk['YEAR'], jk[ ' ANNUAL RAINFALL'], color='Red')
```

### 2018 rainfall is the hight in th

```python
jk.columns
```

```python
data=jk[jk["YEAR"]==2018].squeeze()
month=['FEB','MAR','APR','MAY','JUN','JUL','AUG','SEP','OCT','NOV','DEC']
plt.figure(figsize=(12,6))
plt.bar(month,data[month],color="skyblue",edgecolor="black")
plt.title("Monthly Rainfall In Kerala(2018)",fontsize= 14)
plt.xlabel("Month",fontsize=14)
plt.ylabel("Average Rainfall (mm)",fontsize=14)
plt.show()
```

### heavy rain falls are mostly in jul, aug in 2018

```python
jk['Decade'] = (jk['YEAR'] // 10) * 10
plt.figure(figsize=(11,6))
custom_palette = ['slateblue', 'grey']
sns.countplot(x ='Decade',hue ='FLOODS', data = jk, palette = custom_palette)
plt.title('Flood occurrences by decade', fontsize = 16)
plt.xlabel('Decade')
plt.ylabel('Count')
plt.legend(title = 'Floods')
plt.show()
```

### The graph shows the flood occurrences in a period of 10 years from 1900 to 2010

```python
corr_flood=jk[['YEAR','JAN','FEB','MAR','APR','MAY','JUN','JUL','AUG','SEP','OCT','NOV','DEC',' ANNUAL RAINFALL']].corr()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(corr_flood,annot=True,cbar=True,cmap='Blues')
plt.title("Correlation matrix of numerical values")
plt.show()
```

* From the correlation matrix,rain fall during may,jun,july,august and september are correlated with annual rainfall
* July month shows the highest correlation with the annual rain fall

### conclution
